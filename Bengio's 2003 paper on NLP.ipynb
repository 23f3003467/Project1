```python
import torch
import torch.nn as nn
```

```python
class NNLM(nn.Module):

    def __init__(self, vocab_size, embed_dim, context_size, hidden_dim):

        super().__init__()

        # Embedding matrix
        self.embedding = nn.Embedding(vocab_size, embed_dim)

        # Hidden layer
        self.fc1 = nn.Linear(context_size * embed_dim,
                             hidden_dim)

        # Output layer
        self.fc2 = nn.Linear(hidden_dim,
                             vocab_size)

    def forward(self, x):

        # x shape = (batch_size, context_size)

        x = self.embedding(x)

        # shape
        # (batch_size, context_size, embed_dim)

        x = x.view(x.size(0), -1)

        # concatenate embeddings

        x = torch.tanh(self.fc1(x))

        logits = self.fc2(x)

        return logits
```

```python
model = NNLM(
    vocab_size=10000,
    embed_dim=100,
    context_size=3,
    hidden_dim=300
)
```
